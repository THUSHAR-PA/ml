```python
import pandas as pd
```

```python
df2 = pd.read_csv(r"C:\Users\darks\Downloads\LOTR 2.csv")
```

```python
df1.merge(df2)# inner join
```

```python
df1.merge(df2, how='inner', on='FellowshipID')
```

```python
df1.merge(df2, how = 'outer')
```

```python
df1.merge(df2, how = 'left')
```

```python
df1.merge(df2 , how = 'right')
```

```python
df1.merge(df2 , how = 'cross')# like cartesian product
```
